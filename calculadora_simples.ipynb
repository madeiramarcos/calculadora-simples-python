```python
def sum(a,b):
  soma = a + b
  return soma

def sub(a,b):
  subtracao = a - b
  return subtracao

def mult(a,b):
  multiplicacao = a * b
  return multiplicacao

def div(a,b):
  if b == 0:
    return None
  else:
    divisao = a / b
    return divisao
```

```python
while True:
  print("\n--- CALCULADORA PYTHON ---")
  print("1 - Somar")
  print("2 - Subtrair")
  print("3 - Multiplicar")
  print("4 - Dividir")
  print("0 - Sair")

  opcao = input("Escolha uma opção: ")
  if opcao == "0":
    print("\nCalculadora encerrada")
    break
  elif opcao in ["1","2","3","4"]:
    a = float(input("\nDigite o primeiro número: "))
    b = float(input("Digite o segundo número: "))
    if opcao == "1":
      print(f"\nResultado: {sum(a,b):.2f}")
    elif opcao == "2":
      print(f"\nResultado: {sub(a,b):.2f}")
    elif opcao == "3":
      print(f"\nResultado: {mult(a,b):.2f}")
    elif opcao == "4":
      if  div(a,b) is None:
        print("\nNão é possível dividir por zero")
      else:
        print(f"\nResultado: {div(a,b):.2f}")
  else:
    print("\nOpção inválida")
```
